```python
from datetime import datetime, time

def clean_order_time(value):
    if pd.isna(value):
        return None

    value = str(value).strip()

    # Excel fractional time stored as text
    try:
        numeric_value = float(value)

        if 0 <= numeric_value <= 1:
            total_seconds = round(numeric_value * 24 * 60 * 60)

            hours = (total_seconds // 3600) % 24
            minutes = (total_seconds % 3600) // 60

            return time(hours, minutes)
    except ValueError:
        pass

    # Normal HH:MM format
    try:
        return datetime.strptime(value, "%H:%M").time()
    except ValueError:
        return None


clean_df["Time_Orderd"] = raw_df["Time_Orderd"].apply(clean_order_time)

print("Missing Time_Orderd after cleaning:",
      clean_df["Time_Orderd"].isnull().sum())

print("\nSample cleaned values:")
print(clean_df["Time_Orderd"].dropna().head(20).to_list())
```
